# 일원분산(ANOVA) 분석

```python
import pandas as pd
```

```python
# 데이터 불러오기 및 확인
df = pd.read_csv("datas2/지점별_일별판매량.csv", encoding='euc-kr')
df.head()
```

```python
# 데이터 분할 (결측이 포함되어 있을 수 있으므로 dropna를 사용)
A = df['지점A'].dropna().values
B = df['지점B'].dropna().values
C = df['지점C'].dropna().values
```

### 정규성 검정
- Shapiro-Wilk 검정

```python
from scipy import stats
print(stats.shapiro(A-B))
print(stats.shapiro(A-C))
print(stats.shapiro(B-C))
```

[해석]
- 모두 p-value >= 0.05 : 각 변수간의 차이가, 모두 정규 분포를 뛴다는 것을 알 수 있음.

### 등분산성 검정
- 레빈 검정(Levene's Test) - 권장
- center = 'median'이 기본값이며, 평균(mean)보다 이상치에 덜 민감하여 더 안정적임

```python
levene_stat, levene_p_value = stats.levene(A, B, C, center = 'median')
print(f"levene_stat : {levene_stat}, levene_p_value : {levene_p_value}")
```

[해석]
- p-value >= 0.05, 샘플 간 분산이 유사하다(등분산성을 만족한다)라고 판단할 수 있음

### 일원분산 분석 수행

```python
print(stats.f_oneway(A, B,C))
```

[해석]
- p-value < 0.05 : 거의 0에 가까우므로, 귀무가설을 강력하게 기각함
- 최소 하나의 그룹은 다른 그룹의 평균과 다르다고 판단할 수 있음

### 사후분석

```python
from statsmodels.stats import multicomp 
# 분석 대상 데이터
Data = A.tolist() + B.tolist() + C.tolist()
# 각 값이 속한 그룹 이름
Group = ['A'] * len(A) + ['B'] * len(B) + ['C'] * len(C)

print(Data)
print(Group)
```

```python
# 사후 분석
print(multicomp.pairwise_tukeyhsd(Data, Group))
```

[해석]

- 결론: [A, B] / [C]로 데이터가 구분됨
- reject False, p-adj >= 0.05
    - A - B는 유의한 차이가 존재한다고 보기 어려움
- reject True, p-adj < 0.05
    - A - C는 유의한 차이 존재
- reject True, p-adj < 0.05
    - B - C는 유의한 차이 존재
    - adjusted p-value는 여러 집단을 동시에 비교할 때, 우연히 유의한 결과가 나올 가능성이 커지는 문제를 보정한 p-value
    - "유의한 차이"란 두 집단의 차이가 우연히 생긴 것 같지 않을 정도로 뚜렷하다는 뜻이다.
[tip] reject의 의미

- False : 귀무가설 기각하지 않음.
- True : 귀무가설을 기각함.
