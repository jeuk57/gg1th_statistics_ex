## 정규성 검정하기

```python
# data 불러오기
with open("./datas1/성인여성_키_데이터.txt", "r") as f:
    data = f.read().split("\n")

    data = list(map(float, data))

    print(data)
    print(len(data))

```

```python
from scipy import stats
import numpy as np

mean_val = np.mean(data)
std_dev = np.std(data, ddof=1)  # 표본의 표준편차
```

```python
mean_val, std_dev
```

# 가설 설정
- 귀무가설 : 표본 데이터의 평균은 163과 같다.
- 대립가설 : 표본 데이터의 평균은 163과 다르다.

# 정규성 검정
- 수집한 표본데이터가 정규분포를 따르는지 검증한다.

```python
# stats.kstest(data, 'norm', args=(mean_val, std_dev))\
# 최신 버전에서는 'norm' ->
stats.kstest(
    data,
    stats.norm(
        loc=np.mean(data),
        scale=np.std(data, ddof=0)
    ).cdf
)
```

### 정규성 검정의 가설
- 귀무가설 : 수집한 표본의 데이터는 정규성을 따른다.
- 대립가설 : 수집한 표본의 데이터는 정규성을 따르지 않는다.

```python
# 단일 표본 t 검정 수행
# 검중 수치 = 163, 귀무가설 평균키는 163이다.

print(stats.ttest_1samp(data, 163)) 
```

### 결과 해석
- p-value < 0.05 이므로 -> 귀무가설을 기각한다. 대립가설 채택
- 결과적으로 여성 키의 표본 데이터의 평균은 163과 다르다는 것이 통계적으로 유의미하다.

# 독립 표본 t - 검정

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("./datas2/반별_점수_type1.csv", encoding = "euc-kr")
# df1 = pd.read_csv("./datas2/반별_점수_type1.csv")
df1.head()
```

```python
# A반의 데이터만 추출
# df1['점수'].loc[df1['반'=='A']]
group_A = df1['점수'].loc[df1['반'] == 'A'].values
```

```python
# B반의 데이터만 추출
group_B = df1['점수'].loc[df1['반'] == 'B'].values
```

```python
import numpy as np

# A반의 평균과 표준 편차
mean_A = np.mean(group_A)
std_A = np.std(group_A, ddof=1)  # 표본 표준 편차 (ddof=1)

# B반의 평균과 표준 편차
mean_B = np.mean(group_B)
std_B = np.std(group_B, ddof=1)  # 표본 표준 편차 (ddof=1)

# 결과 출력
print(f"A반 평균: {mean_A:.2f}, 표준 편차: {std_A:.2f}")
print(f"B반 평균: {mean_B:.2f}, 표준 편차: {std_B:.2f}")
```

## 정규성 검정
- 정규성 검정 가설수립
    - 영가설 : 두 그룹의 표본 데이터는 정규분포를 따른다.
    - 대립가설 : 두 그룹의 표본 데이터는 정규분포를 따르지 않는다.

```python
print(stats.kstest(
  group_A, 
  stats.norm.cdf, 
  args=(np.mean(group_A), np.std(group_A, ddof=1))))
```

```python
print(stats.kstest(
  group_B, 
  stats.norm.cdf, 
  args=(np.mean(group_B), np.std(group_B, ddof=1))))
```

[해석]
- p-value >= 0.05 : 귀무가설을 기각하지 않는다. 정규성을 따른다고 볼 수 있다.

## 등분산성 검정
- 가설 수립
    - 영가설(귀무가설) : 두 그룹은 분산이 유사하다.(등분산을 따른다.)
    - 대립가설 : 두 그룹은 분산이 유사하지 않다.(등분산을 따르지 않는다.)

```python
stats.levene(group_A, group_B)
```

[해석]
p-value >= 0.005 : 두 그룹은 등분산성을 따른다.

##  독립 표본 t - 검정

- 가설 수립
    - 영가설(귀무가설) : 두 반의 표본의 평균은 같다.
    - 대립가설 : 두 반의 표본의 평균은 다르다.

```python
stats.ttest_ind(group_A, group_B, equal_var = True)
```

[해석]
- p-value < 0.05 : 귀무가설을 기각한다. 대립가설 채택.
- 두 반의 표본의 평균은 다르다는 것이 통계적으로 유의하다.
